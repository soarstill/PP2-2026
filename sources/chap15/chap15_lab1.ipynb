```python
import pandas as pd
```

```python
df = pd.read_csv("shopping.csv")
```

```python
print("=== 앞부분 미리 보기 ===")
print(df.head())
```

```python
print("\n=== 전체 정보 ===")
print(df.info())
```

```python
print("\n=== 숫자형 컬럼 통계 ===")
print(df.describe())
```

```python
print("\n=== 열별 결측치 개수 ===")
print(df.isna().sum())
```

```python
print("\nUnitPrice 통계:")
print(df["UnitPrice"].describe())
```

```python
# 주문별 매출(단가 × 수량)
df["Revenue"] = df["UnitPrice"] * df["Quantity"]
print("\n전체 매출 합계:", df["Revenue"].sum())
```

```python
# 서울 지역만 필터링
seoul_df = df[df["Region"] == "Seoul"]
print("\n서울 지역 주문 수:", len(seoul_df))
print("서울 지역 매출 합계:", seoul_df["Revenue"].sum())
```
