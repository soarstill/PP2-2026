## GUI 프로그래밍
- tkinter 모듈(내장GUI라이브러리)을 이용한 GUI 프로그래밍 원리를 배운다.
- 파이썬 내장모듈이므로 별도의 설치과정은 없다.
- 그외의 GUI 라이브러리 : Pyside6 (PyQt) 





````python

import tkinter as tk
from tkinter import ttk, messagebox

# 버튼 클릭 시 호출되는 콜백: 눌린 버튼의 이름을 메시지 박스로 보여 줍니다.
def on_button_click(name):
    messagebox.showinfo("알림", f"'{name}' 버튼이 눌렸습니다.")

# 1. 메인 윈도우 설정
root = tk.Tk()
root.title("Widgets Test")
root.geometry("620x720")
root.configure(bg="#CECAD3") # 클래식 윈도우 회색 배경

# --- 메뉴바 (Menu) ---
menubar = tk.Menu(root)
filemenu = tk.Menu(menubar, tearoff=0)
filemenu.add_command(label="Exit", command=root.quit)
menubar.add_cascade(label="File", menu=filemenu)
root.config(menu=menubar)

# 전체적인 패딩 스타일 정의
pad_opts = {'padx': 10, 'pady': 6}

# --- 1행: Button & Checkboxes ---
lbl_button = tk.Label(root, text="Button:", bg="#D4D0C8", anchor="w")
lbl_button.grid(row=0, column=0, sticky="w", **pad_opts)

# command에 lambda를 써서 이 버튼의 이름을 콜백에 전달합니다.
btn_sample = tk.Button(root, text="Button", width=12, relief="raised", command=lambda: on_button_click("Button"))
btn_sample.grid(row=0, column=1, sticky="w", **pad_opts)

# 우측 체크박스 그룹용 프레임
chk_frame = tk.Frame(root, bg="#D4D0C8")
chk_frame.grid(row=0, column=2, rowspan=3, sticky="nw", padx=30, pady=5)

chk_enabled = tk.Checkbutton(chk_frame, text="Enabled", bg="#D4D0C8")
chk_enabled.select() # 체크 활성화
chk_enabled.pack(anchor="w", pady=2)

chk_visible = tk.Checkbutton(chk_frame, text="Visible", bg="#D4D0C8")
chk_visible.select()
chk_visible.pack(anchor="w", pady=2)

chk_editable = tk.Checkbutton(chk_frame, text="Editable", bg="#D4D0C8")
chk_editable.select()
chk_editable.pack(anchor="w", pady=2)

# --- 2행: TextField ---
lbl_textfield = tk.Label(root, text="TextField:", bg="#D4D0C8", anchor="w")
lbl_textfield.grid(row=1, column=0, sticky="w", **pad_opts)

entry_text = tk.Entry(root, width=30, relief="sunken")
entry_text.grid(row=1, column=1, sticky="w", **pad_opts)

# --- 3행: PasswordField ---
lbl_password = tk.Label(root, text="PasswordField:", bg="#D4D0C8", anchor="w")
lbl_password.grid(row=2, column=0, sticky="w", **pad_opts)

entry_pass = tk.Entry(root, width=30, show="*", relief="sunken")
entry_pass.grid(row=2, column=1, sticky="w", **pad_opts)

# --- 4행: TextArea & 속성 버튼들 ---
lbl_textarea = tk.Label(root, text="TextArea:", bg="#D4D0C8", anchor="w")
lbl_textarea.grid(row=3, column=0, sticky="nw", **pad_opts)

# 스크롤바가 포함된 텍스트 영역
txt_frame = tk.Frame(root)
txt_frame.grid(row=3, column=1, sticky="w", **pad_opts)

scrollbar = tk.Scrollbar(txt_frame)
scrollbar.pack(side="right", fill="y")

text_area = tk.Text(txt_frame, width=28, height=7, relief="sunken", yscrollcommand=scrollbar.set)
text_area.insert("1.0", "Use the checkboxes and\nbuttons on the right to set the\nattributes of the widgets on the\nleft.\n\nThe font support is not hooked\nup yet.")
text_area.pack(side="left", fill="both")
scrollbar.config(command=text_area.yview)

# 우측 속성 제어 버튼들
btn_frame = tk.Frame(root, bg="#D4D0C8")
btn_frame.grid(row=3, column=2, sticky="nw", padx=30, pady=5)

buttons = ["BackgroundColor", "ForegroundColor", "Font", "ToolTip", "Background BackgroundColor"]
for btn_text in buttons:
    # t=btn_text 기본값으로 현재 반복의 이름을 고정해야 모든 버튼이 마지막 이름을 보여 주는 문제를 피합니다.
    b = tk.Button(btn_frame, text=btn_text, width=22, relief="raised", command=lambda t=btn_text: on_button_click(t))
    b.pack(pady=3, anchor="w")

# --- 5행: StaticText ---
lbl_static_lbl = tk.Label(root, text="StaticText:", bg="#D4D0C8", anchor="w")
lbl_static_lbl.grid(row=4, column=0, sticky="w", **pad_opts)

lbl_static_val = tk.Label(root, text="StaticText", bg="#D4D0C8")
lbl_static_val.grid(row=4, column=1, sticky="w", **pad_opts)

# --- 6행: CheckBox ---
lbl_checkbox_lbl = tk.Label(root, text="CheckBox:", bg="#D4D0C8", anchor="w")
lbl_checkbox_lbl.grid(row=5, column=0, sticky="w", **pad_opts)

chk_sample = tk.Checkbutton(root, text="CheckBox", bg="#D4D0C8")
chk_sample.grid(row=5, column=1, sticky="w", **pad_opts)

# --- 7행: RadioGroup & Image 영역 ---
lbl_radiogroup = tk.Label(root, text="RadioGroup:", bg="#D4D0C8", anchor="w")
lbl_radiogroup.grid(row=6, column=0, sticky="nw", **pad_opts)

# 테두리가 있는 라디오박스 프레임
radio_frame = tk.LabelFrame(root, text="A RadioBox", bg="#D4D0C8", padx=10, pady=5)
radio_frame.grid(row=6, column=1, sticky="w", **pad_opts)

radio_var = tk.StringVar(value="one")
r1 = tk.Radiobutton(radio_frame, text="one", value="one", variable=radio_var, bg="#D4D0C8")
r2 = tk.Radiobutton(radio_frame, text="two", value="two", variable=radio_var, bg="#D4D0C8")
r3 = tk.Radiobutton(radio_frame, text="three", value="three", variable=radio_var, bg="#D4D0C8")
r1.pack(anchor="w")
r2.pack(anchor="w")
r3.pack(anchor="w")

# 우측 Image 캔버스 영역
lbl_image = tk.Label(root, text="Image:", bg="#D4D0C8")
lbl_image.grid(row=6, column=2, sticky="nw", padx=30, pady=5)

# 격자 무늬 배경의 더미 이미지 공간 (Canvas 사용)
canvas_img = tk.Canvas(root, width=100, height=100, bg="#A0B0C0", highlightthickness=1, highlightbackground="gray")
canvas_img.grid(row=6, column=2, sticky="nw", padx=90, pady=5)
# 모눈종이 같은 격자선 그리기
for i in range(0, 100, 10):
    canvas_img.create_line(i, 0, i, 100, fill="#B0C0D0")
    canvas_img.create_line(0, i, 100, i, fill="#B0C0D0")

# --- 8행: Choice & ImageButton ---
lbl_choice = tk.Label(root, text="Choice:", bg="#D4D0C8", anchor="w")
lbl_choice.grid(row=7, column=0, sticky="w", **pad_opts)

# Choice에 대응하는 ttk.Combobox
combo_choice = ttk.Combobox(root, values=["one", "two", "three"], width=8, state="readonly")
combo_choice.set("two")
combo_choice.grid(row=7, column=1, sticky="w", **pad_opts)

lbl_imgbtn = tk.Label(root, text="ImageButton:", bg="#D4D0C8")
lbl_imgbtn.grid(row=7, column=2, sticky="w", padx=30, pady=5)

btn_edit = tk.Button(root, text="📝 edit", width=8, bg="#6A6A9B", fg="white", relief="raised", command=lambda: on_button_click("edit"))
btn_edit.grid(row=7, column=2, sticky="w", padx=120, pady=5)

# --- 9행: List & Borderless TextField ---
lbl_list = tk.Label(root, text="List:", bg="#D4D0C8", anchor="w")
lbl_list.grid(row=8, column=0, sticky="nw", **pad_opts)

listbox = tk.Listbox(root, width=10, height=4, relief="sunken")
for item in ["one", "two", "three"]:
    listbox.insert("end", item)
listbox.select_set(2) # 'three' 선택 상태로 만들기
listbox.grid(row=8, column=1, sticky="w", **pad_opts)

lbl_noborder = tk.Label(root, text="TextField with no border", bg="#D4D0C8")
lbl_noborder.grid(row=8, column=2, sticky="nw", padx=30, pady=5)

# --- 10행: Slider (Scale) ---
lbl_slider = tk.Label(root, text="Slider:", bg="#D4D0C8", anchor="w")
lbl_slider.grid(row=9, column=0, sticky="w", **pad_opts)

slider = tk.Scale(root, from_=0, to=100, orient="horizontal", showvalue=False, width=10, length=200, bg="#D4D0C8", bd=1)
slider.set(20)
slider.grid(row=9, column=1, columnspan=2, sticky="w", **pad_opts)

# --- 11행: StaticLine ---
lbl_line = tk.Label(root, text="StaticLine:", bg="#D4D0C8", anchor="w")
lbl_line.grid(row=10, column=0, sticky="w", **pad_opts)

# 구분선 구현 (두께 2픽셀의 프레임)
sep_line = tk.Frame(root, height=2, bd=1, relief="sunken", bg="gray")
sep_line.grid(row=10, column=1, columnspan=2, sticky="we", padx=10, pady=10)

# --- 클릭 시 위젯 이름을 풍선말로 표시 ---
# 변수 이름으로 위젯을 찾을 수 있도록 {위젯 경로: 변수 이름} 사전을 만듭니다.
var_names = {str(v): k for k, v in list(globals().items()) if isinstance(v, tk.Widget)}
tip = {'win': None, 'job': None}  # 현재 떠 있는 풍선과 자동 닫기 예약을 저장합니다.

def hide_tip():
    if tip['job']:
        root.after_cancel(tip['job'])  # 예약된 닫기 작업을 취소합니다.
        tip['job'] = None
    if tip['win']:
        tip['win'].destroy()  # 기존 풍선을 제거합니다.
        tip['win'] = None

def show_tip(event):
    w = event.widget
    name = var_names.get(str(w), w.winfo_class())  # 변수 이름, 없으면 위젯 종류
    if 'text' in w.keys() and w.cget('text'):
        name += f" ({w.cget('text')})"  # 버튼·라벨 등은 표시 글자도 함께 보여 줍니다.
    hide_tip()
    win = tk.Toplevel(root)
    win.wm_overrideredirect(True)  # 제목 표시줄 없는 풍선 창으로 만듭니다.
    win.wm_geometry(f"+{event.x_root + 12}+{event.y_root + 12}")  # 마우스 오른쪽 아래에 표시합니다.
    tk.Label(win, text=name, bg="#FFFFE0", relief="solid", bd=1, padx=6, pady=2).pack()
    tip['win'] = win
    tip['job'] = root.after(1500, hide_tip)  # 1.5초 뒤 자동으로 사라집니다.

def bind_tips(widget):
    widget.bind('<Button-1>', show_tip, add='+')  # add='+'로 기존 클릭 동작을 유지합니다.
    for child in widget.winfo_children():
        bind_tips(child)  # 프레임 안의 위젯까지 재귀적으로 연결합니다.

for child in root.winfo_children():
    bind_tips(child)

# 메인 루프 실행
root.mainloop()
````

## Tkinter 위젯 코드와 상세 해설

아래는 기존 화면 예제를 그대로 유지하면서 위젯의 역할과 주요 설정을 주석으로 설명한 코드입니다. 이 셀은 설명용 Markdown 셀이므로, 실행하려면 코드 블록 안을 Python 코드 셀에 복사하세요.

```python
import tkinter as tk
from tkinter import ttk

# Tkinter로 윈도우와 화면 요소(위젯)를 만듭니다.

# 1. 메인 윈도우 설정
root = tk.Tk()  # 모든 위젯이 들어가는 최상위 창입니다.
root.title("Widgets Test")  # 창 제목 표시줄의 이름을 정합니다.
root.geometry("620x720")  # 창의 초기 크기를 너비 620, 높이 720픽셀로 설정합니다.
root.configure(bg="#D4D0C8")  # 창 배경을 클래식 윈도우 스타일의 회색으로 설정합니다.

# --- 메뉴바 (Menu) ---
menubar = tk.Menu(root)  # 메인 창에 연결할 메뉴 표시줄을 만듭니다.
filemenu = tk.Menu(menubar, tearoff=0)  # File 메뉴를 만들며, 메뉴를 떼어내는 기능은 끕니다.
filemenu.add_command(label="Exit", command=root.quit)  # Exit를 누르면 이벤트 루프를 종료합니다.
menubar.add_cascade(label="File", menu=filemenu)  # 메뉴 표시줄에 File 메뉴를 추가합니다.
root.config(menu=menubar)  # 메뉴 표시줄을 메인 창에 표시합니다.

# 위젯의 바깥쪽 여백을 공통으로 지정합니다.
# padx는 좌우 여백, pady는 위아래 여백이며 **pad_opts로 grid에 전달합니다.
pad_opts = {'padx': 10, 'pady': 6}

# --- 1행: Button & Checkboxes ---
# Label은 위젯 이름이나 안내 문구를 표시합니다.
lbl_button = tk.Label(root, text="Button:", bg="#D4D0C8", anchor="w")
lbl_button.grid(row=0, column=0, sticky="w", **pad_opts)  # grid의 0행 0열에 놓고 왼쪽 정렬합니다.

row=0, column=0: 0행 0열(위에서 두 번째 줄, 맨 왼쪽 칸)에 배치합니다. 번호는 0부터 시작합니다.
sticky="w": 셀 안에서 위젯을 왼쪽(west)에 붙입니다. 이 줄은 "TextField:" 라벨 위치입니다.
'''
**pad_opts: 딕셔너리 {'padx': 10, 'pady': 6}를 풀어서 padx=10, pady=6으로 전달합니다. 위젯 바깥의 좌우 10픽셀, 위아래 6픽셀 여백입니다.
**는 딕셔너리를 키워드 인자로 펼치는 파이썬 문법이라, 같은 여백을 여러 위젯에 반복해서 쓸 수 있습니다.
'''
# Button은 클릭 가능한 버튼입니다. command가 없으므로 현재는 별도 동작을 하지 않습니다.
btn_sample = tk.Button(root, text="Button", width=12, relief="raised")
btn_sample.grid(row=0, column=1, sticky="w", **pad_opts)  # 0행 1열에 배치합니다.

# Frame은 관련 위젯을 하나의 그룹으로 묶는 컨테이너입니다.
chk_frame = tk.Frame(root, bg="#D4D0C8")
chk_frame.grid(row=0, column=2, rowspan=3, sticky="nw", padx=30, pady=5)  # 2열에서 3개 행을 차지하며 왼쪽 위에 붙입니다.
'''
row=0: 0번 행, 즉 첫 번째 행에 배치합니다. 행 번호는 0부터 시작합니다.
column=2: 2번 열, 즉 세 번째 열에 배치합니다.
rowspan=3: 현재 위치에서 세 행을 차지합니다. 여기서는 0~2번 행에 걸칩니다.
sticky="nw": 차지하는 영역 안에서 위쪽(n)과 왼쪽(w)에 붙입니다.
padx=30: 위젯 바깥에 좌우 여백을 각각 30픽셀 둡니다.
pady=5: 위젯 바깥 상하 여백을 각각 5픽셀 둡니다.
'''


# Checkbutton은 선택 또는 해제 상태를 표시하는 체크박스입니다.
chk_enabled = tk.Checkbutton(chk_frame, text="Enabled", bg="#D4D0C8")
chk_enabled.select()  # 처음부터 체크된 상태로 표시합니다.
chk_enabled.pack(anchor="w", pady=2)  # 프레임 안에서 왼쪽 정렬하고 위아래 간격을 둡니다.
'''
anchor는 위젯 안에서 내용(텍스트 등)을 어느 쪽에 붙여 정렬할지 정하는 옵션입니다. 값은 방위(n, s, e, w와 조합)로 지정합니다.

값: "center"(기본값, 가운데), "w"(왼쪽), "e"(오른쪽), "n"(위), "s"(아래), "nw"(왼쪽 위), "ne"(오른쪽 위), "sw"(왼쪽 아래), "se"(오른쪽 아래)

이 코드에서는 두 가지 방식으로 쓰였습니다.

Label의 anchor="w"


라벨 안의 글자를 라벨 영역의 왼쪽에 붙입니다. 라벨이 글자보다 넓을 때 효과가 보입니다.

pack(anchor="w")


pack으로 쌓을 때, 부모 프레임이라는 공간 안에서 위젯 자체를 왼쪽에 붙입니다. 이 옵션이 없으면 가운데에 놓입니다.

grid의 sticky와 비슷해 보이지만 대상이 다릅니다.

'''
chk_visible = tk.Checkbutton(chk_frame, text="Visible", bg="#D4D0C8")
chk_visible.select()  # 두 번째 체크박스도 초기 선택 상태로 설정합니다.
chk_visible.pack(anchor="w", pady=2)  # 왼쪽 정렬, 상하 2픽셀 여백

chk_editable = tk.Checkbutton(chk_frame, text="Editable", bg="#D4D0C8")
chk_editable.select()  # 세 번째 체크박스도 초기 선택 상태로 설정합니다.
chk_editable.pack(anchor="w", pady=2)

# --- 2행: TextField ---
# Entry는 한 줄의 문자열을 입력받는 위젯입니다.
lbl_textfield = tk.Label(root, text="TextField:", bg="#D4D0C8", anchor="w")
lbl_textfield.grid(row=1, column=0, sticky="w", **pad_opts) # 1행, 0열, 

entry_text = tk.Entry(root, width=30, relief="sunken")  # width는 표시 폭, relief는 테두리 모양입니다.
entry_text.grid(row=1, column=1, sticky="w", **pad_opts)

# --- 3행: PasswordField ---
# show="*"를 지정해 입력한 문자를 별표로 가리는 비밀번호 입력 칸입니다.
lbl_password = tk.Label(root, text="PasswordField:", bg="#D4D0C8", anchor="w")
lbl_password.grid(row=2, column=0, sticky="w", **pad_opts)

entry_pass = tk.Entry(root, width=30, show="*", relief="sunken")
entry_pass.grid(row=2, column=1, sticky="w", **pad_opts)
'''
root 창 안에 한 줄 입력 위젯(Entry)을 만들어 entry_pass에 저장합니다.

tk.Entry(root, ...): 한 줄짜리 텍스트 입력 위젯이며, 부모는 root(메인 창)입니다.
width=30: 입력 칸의 폭을 글자 30개 정도로 정합니다. 픽셀이 아니라 <글자 수>단위입니다.
show="*": 입력한 글자를 화면에는 *로 대신 표시합니다. 값은 그대로 저장되고 보이는 모양만 바뀌므로, 비밀번호 입력에 씁니다.
relief="sunken": 테두리를 움푹 들어간 모양으로 그립니다. 입력 칸에서 흔히 쓰는 스타일입니다. 
 다른 값으로는 raised(튀어나옴), flat(평평), groove, ridge, solid가 있습니다.
  입력한 실제 값은 entry_pass.get()으로 읽을 수 있습니다. 
'''

# --- 4행: TextArea & 속성 버튼들 ---
# Text는 여러 줄의 텍스트를 표시하거나 입력하는 위젯입니다.
lbl_textarea = tk.Label(root, text="TextArea:", bg="#D4D0C8", anchor="w")
lbl_textarea.grid(row=3, column=0, sticky="nw", **pad_opts)  # 여러 줄 영역의 위쪽에 맞춰 배치합니다.

# 텍스트 영역과 스크롤바를 함께 담을 프레임입니다.
txt_frame = tk.Frame(root)
txt_frame.grid(row=3, column=1, sticky="w", **pad_opts)
'''
root 창 안에 빈 Frame(프레임) 을 만드는 코드입니다. 프레임은 다른 위젯을 담는 보이지 않는 상자(컨테이너) 로, 관련 위젯을 한 묶음으로 다룰 때 씁니다.

tk.Frame(root): 부모가 root인 프레임을 만들고 txt_frame에 저장합니다. 배경색이나 크기 옵션이 없어 기본 모양입니다.
'''
scrollbar = tk.Scrollbar(txt_frame)  # 세로 스크롤바를 만듭니다.
scrollbar.pack(side="right", fill="y")  # 프레임 오른쪽에 놓고 세로 방향으로 채웁니다.

# yscrollcommand는 텍스트의 현재 스크롤 위치를 스크롤바에 반영합니다.
text_area = tk.Text(txt_frame, width=28, height=7, relief="sunken", yscrollcommand=scrollbar.set)
'''
tk.Text(txt_frame, ...): 부모는 txt_frame입니다. 메인 창이 아니라 앞에서 만든 프레임 안에 들어갑니다.
width=28: 폭을 글자 28개 정도로 정합니다.
height=7: 높이를 7줄로 정합니다. 글자가 이보다 많아지면 스크롤이 필요합니다.
relief="sunken": 테두리를 움푹 들어간 모양으로 그립니다.
yscrollcommand=scrollbar.set: 텍스트가 스크롤될 때마다 스크롤바에 현재 위치를 알려주는 연결입니다. 내용이 움직이면 scrollbar.set이 호출되어 스크롤바의 손잡이 위치와 크기가 갱신됩니다. 여기서는 함수를 호출하지 않고 set이라는 함수 자체를 넘기므로 괄호가 없습니다.
'''

# Text의 인덱스 "1.0"은 첫 번째 줄의 첫 번째 문자 위치입니다.
text_area.insert("1.0", "Use the checkboxes and\nbuttons on the right to set the\nattributes of the widgets on the\nleft.\n\nThe font support is not hooked\nup yet.")
text_area.pack(side="left", fill="both")  # 스크롤바 왼쪽에 두고 프레임 공간에 맞춰 늘립니다.
scrollbar.config(command=text_area.yview)  # 스크롤바를 움직이면 Text의 세로 보기가 이동하도록 연결합니다.

# 오른쪽 속성 버튼들을 모아 둘 프레임입니다.
btn_frame = tk.Frame(root, bg="#D4D0C8")
btn_frame.grid(row=3, column=2, sticky="nw", padx=30, pady=5)

# 버튼 문구를 리스트로 관리하면 반복문으로 같은 형식의 버튼을 만들 수 있습니다.
buttons = ["BackgroundColor", "ForegroundColor", "Font", "ToolTip", "Background BackgroundColor"]
for btn_text in buttons:  # 리스트의 문구를 하나씩 꺼냅니다.
    b = tk.Button(btn_frame, text=btn_text, width=22, relief="raised")  # 현재 문구로 버튼을 만듭니다.
    b.pack(pady=3, anchor="w")  # 버튼 사이 간격을 두고 왼쪽부터 세로로 쌓습니다.

# --- 5행: StaticText ---
# StaticText는 사용자가 편집하지 않는 표시용 텍스트를 뜻합니다.
lbl_static_lbl = tk.Label(root, text="StaticText:", bg="#D4D0C8", anchor="w")
lbl_static_lbl.grid(row=4, column=0, sticky="w", **pad_opts)

lbl_static_val = tk.Label(root, text="StaticText", bg="#D4D0C8")
lbl_static_val.grid(row=4, column=1, sticky="w", **pad_opts)

# --- 6행: CheckBox ---
# 앞의 그룹과 별도로 메인 창에 체크박스를 하나 더 만듭니다.
lbl_checkbox_lbl = tk.Label(root, text="CheckBox:", bg="#D4D0C8", anchor="w")
lbl_checkbox_lbl.grid(row=5, column=0, sticky="w", **pad_opts)

chk_sample = tk.Checkbutton(root, text="CheckBox", bg="#D4D0C8")
chk_sample.grid(row=5, column=1, sticky="w", **pad_opts)

# --- 7행: RadioGroup & Image 영역 ---
# 같은 variable을 공유하는 Radiobutton은 그룹 안에서 하나만 선택할 수 있습니다.
lbl_radiogroup = tk.Label(root, text="RadioGroup:", bg="#D4D0C8", anchor="w")
lbl_radiogroup.grid(row=6, column=0, sticky="nw", **pad_opts)

# LabelFrame은 제목과 테두리가 있는 그룹 컨테이너입니다.
radio_frame = tk.LabelFrame(root, text="A RadioBox", bg="#D4D0C8", padx=10, pady=5)
radio_frame.grid(row=6, column=1, sticky="w", **pad_opts)

# StringVar에 그룹의 현재 선택값을 저장하며, 처음에는 "one"을 선택합니다.
radio_var = tk.StringVar(value="one")
r1 = tk.Radiobutton(radio_frame, text="one", value="one", variable=radio_var, bg="#D4D0C8")
r2 = tk.Radiobutton(radio_frame, text="two", value="two", variable=radio_var, bg="#D4D0C8")
r3 = tk.Radiobutton(radio_frame, text="three", value="three", variable=radio_var, bg="#D4D0C8")
r1.pack(anchor="w")  # 세 항목을 프레임 안에 세로로 쌓고 왼쪽 정렬합니다.
r2.pack(anchor="w")
r3.pack(anchor="w")

# 실제 이미지 대신 격자 그림을 표시할 Canvas 영역의 제목입니다.
lbl_image = tk.Label(root, text="Image:", bg="#D4D0C8")
lbl_image.grid(row=6, column=2, sticky="nw", padx=30, pady=5)

# Canvas는 선, 도형, 이미지 등을 직접 그릴 수 있는 영역입니다.
canvas_img = tk.Canvas(root, width=100, height=100, bg="#A0B0C0", highlightthickness=1, highlightbackground="gray")
canvas_img.grid(row=6, column=2, sticky="nw", padx=90, pady=5)
# 10픽셀 간격으로 세로선과 가로선을 그려 모눈종이처럼 표시합니다.
for i in range(0, 100, 10):
    canvas_img.create_line(i, 0, i, 100, fill="#B0C0D0")  # x좌표 i를 따라 위에서 아래로 세로선을 그립니다.
    canvas_img.create_line(0, i, 100, i, fill="#B0C0D0")  # y좌표 i를 따라 왼쪽에서 오른쪽으로 가로선을 그립니다.

# --- 8행: Choice & ImageButton ---
# Choice는 여러 선택지 중 하나를 고르는 콤보박스로 구현합니다.
lbl_choice = tk.Label(root, text="Choice:", bg="#D4D0C8", anchor="w")
lbl_choice.grid(row=7, column=0, sticky="w", **pad_opts)

# ttk.Combobox는 드롭다운 목록입니다. readonly이면 목록의 값만 선택할 수 있습니다.
combo_choice = ttk.Combobox(root, values=["one", "two", "three"], width=8, state="readonly")
combo_choice.set("two")  # 처음 표시할 선택값을 "two"로 지정합니다.
combo_choice.grid(row=7, column=1, sticky="w", **pad_opts)

# 이미지 파일 대신 이모지와 텍스트로 이미지 버튼 모양을 표현한 예시입니다.
lbl_imgbtn = tk.Label(root, text="ImageButton:", bg="#D4D0C8")
lbl_imgbtn.grid(row=7, column=2, sticky="w", padx=30, pady=5)

btn_edit = tk.Button(root, text="📝 edit", width=8, bg="#6A6A9B", fg="white", relief="raised")
btn_edit.grid(row=7, column=2, sticky="w", padx=120, pady=5)

# --- 9행: List & Borderless TextField ---
# Listbox는 여러 문자열을 목록으로 보여 주고 항목을 선택할 수 있습니다.
lbl_list = tk.Label(root, text="List:", bg="#D4D0C8", anchor="w")
lbl_list.grid(row=8, column=0, sticky="nw", **pad_opts)

listbox = tk.Listbox(root, width=10, height=4, relief="sunken")
for item in ["one", "two", "three"]:
    listbox.insert("end", item)  # 각 항목을 목록의 마지막에 추가합니다.
listbox.select_set(2)  # 인덱스는 0부터 시작하므로 세 번째 항목인 "three"를 선택합니다.
listbox.grid(row=8, column=1, sticky="w", **pad_opts)

# 테두리 없는 입력 칸 예시를 설명하는 라벨입니다.
lbl_noborder = tk.Label(root, text="TextField with no border", bg="#D4D0C8")
lbl_noborder.grid(row=8, column=2, sticky="nw", padx=30, pady=5)

# --- 10행: Slider (Scale) ---
# Scale은 일정 범위 안에서 값을 움직여 선택하는 슬라이더입니다.
lbl_slider = tk.Label(root, text="Slider:", bg="#D4D0C8", anchor="w")
lbl_slider.grid(row=9, column=0, sticky="w", **pad_opts)

# from_와 to는 최솟값/최댓값, orient는 방향을 설정합니다.
# showvalue=False는 숫자를 숨기며, length는 슬라이더 길이를 지정합니다.
slider = tk.Scale(root, from_=0, to=100, orient="horizontal", showvalue=False, width=10, length=200, bg="#D4D0C8", bd=1)
slider.set(20)  # 처음 표시할 값을 20으로 지정합니다.
slider.grid(row=9, column=1, columnspan=2, sticky="w", **pad_opts)  # columnspan=2는 두 열에 걸쳐 배치한다는 뜻입니다.

# --- 11행: StaticLine ---
# StaticLine은 화면에서 영역을 시각적으로 나누는 구분선입니다.
lbl_line = tk.Label(root, text="StaticLine:", bg="#D4D0C8", anchor="w")
lbl_line.grid(row=10, column=0, sticky="w", **pad_opts)

# 얇고 오목한 테두리의 Frame을 이용해 가로 구분선을 만듭니다.
sep_line = tk.Frame(root, height=2, bd=1, relief="sunken", bg="gray")
sep_line.grid(row=10, column=1, columnspan=2, sticky="we", padx=10, pady=10)  # 두 열에 걸쳐 가로로 늘립니다.

# mainloop()는 사용자 입력과 창 이벤트를 계속 처리하는 이벤트 루프입니다.
# 이 루프가 실행되는 동안 창이 유지되고, 창을 닫으면 GUI 처리가 끝납니다.
root.mainloop()
```

### 배치와 옵션 해설
- `grid(row, column)`: 위젯을 행과 열로 이루어진 격자에 배치합니다.
- `pack(...)`: 위젯을 컨테이너 안에 순서대로 쌓아 배치합니다.
- `sticky="w"` / `sticky="nw"`: 격자 셀 안에서 왼쪽 / 왼쪽 위에 붙입니다.
- `rowspan` / `columnspan`: 위젯이 차지할 행 / 열의 개수를 지정합니다.
- `padx` / `pady`: 위젯 주변의 가로 / 세로 여백입니다.
